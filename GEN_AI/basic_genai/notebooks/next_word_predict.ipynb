## Step 1: Prepare training text
This sample paragraph provides the text the model will learn from.

```python
text="""John woke up early in the morning and looked outside the window. 
The sky was clear, and the streets were quiet. 
He decided to make a cup of coffee before starting his work. 
After breakfast, John opened his laptop and checked his emails. One message was from his friend David, who invited him to visit a new restaurant in the city. 
John replied that he would be happy to join him later in the evening."""
```

## Step 2: Count words
`split()` separates the text into words, and `len()` counts the total number of words.

```python
len(text.split())
```

## Step 3: Install TensorFlow
TensorFlow provides the machine-learning tools required to build and train the model.

```python
!pip install tensorflow
```

## Step 4: Import preprocessing tools
`Tokenizer` converts words into numbers. `pad_sequences` makes sequences the same length.

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

## Step 5: Create the tokenizer
A tokenizer creates a dictionary that maps each word to a unique number.

```python
tokenizer = Tokenizer()
```

## Step 6: Build the vocabulary
`fit_on_texts()` reads the training text and creates the word-to-index mapping.

```python
tokenizer.fit_on_texts([text])
```

## Step 7: View the vocabulary
`word_index` displays the numeric index assigned to every word.

```python
tokenizer.word_index
```

## Step 8: Split the text
The text is divided into separate lines or sentences for processing.

```python
text_split=text.split("\n")
```

## Step 9: Display the split text
This shows the individual text sections created in the previous step.

```python
text_split
```

## Step 10: Convert sentences to numbers
Each word in a sentence is replaced with its tokenizer index.

```python
for sentence in text_split:
    print(tokenizer.texts_to_sequences([sentence])[0])
```

## Step 11: Create n-gram training sequences
Each sequence contains previous words followed by the word the model should predict.

```python
input_sequences = []

for sentence in text_split:
    token_list = tokenizer.texts_to_sequences([sentence])[0]
    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i+1]
        input_sequences.append(n_gram_sequence)
```

## Step 12: Display training sequences
This displays the generated input sequences and their target words.

```python
input_sequences
```

## Step 13: Find the longest sequence
The maximum sequence length is required to pad all sequences consistently.

```python
max_len = max([len(x) for x in input_sequences])
```

## Step 14: Display the maximum length
This shows the length of the longest generated sequence.

```python
max_len
```

## Step 15: Pad the sequences
Zeros are added before shorter sequences so every sequence has the same length.

```python
padded_sequences = pad_sequences(input_sequences, padding='pre', maxlen=max_len)
```

## Step 16: Display padded sequences
This displays the sequences after padding has been applied.

```python
padded_sequences
```

## Step 17: Create input data
`x` contains all words except the final word in each sequence.

```python
x = padded_sequences[:,:-1]
```

## Step 18: Create target data
`y` contains the final word that the model must learn to predict.

```python
y = padded_sequences[:,-1]
```

## Step 19: View input data
This displays the numeric input sequences used for training.

```python
x
```

## Step 20: View target data
This displays the expected next-word indexes.

```python
y
```

## Step 21: Check input dimensions
`shape` shows the number of training examples and input sequence length.

```python
x.shape
```

## Step 22: One-hot encode the targets
`to_categorical()` converts each target index into a one-hot encoded vector.

```python
from tensorflow.keras.utils import to_categorical
y=to_categorical(y, num_classes=56)
```

## Step 23: Inspect an encoded target
This displays the one-hot vector for the first target word.

```python
y[0]
```

## Step 24: Import model components
These Keras classes are used to create the neural-network model.

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Bidirectional
model = Sequential()
```

## Step 25: Build the model layers
The embedding layer represents words as vectors, the LSTM learns word order, and the Dense layer predicts the next word.

```python
model.add(Embedding(input_dim=56, output_dim=10, input_length=max_len-1))
model.add(LSTM(150))
model.add(Dense(56, activation='softmax'))
```

## Step 26: Compile the model
The loss function measures prediction errors, while Adam updates the model weights.

```python
model.compile(loss="categorical_crossentropy", optimizer="adam", metrics=['accuracy'])
```

## Step 27: Build and summarize the model
This initializes the model and displays its layers and number of parameters.

```python
model.build((None, max_len-1))
model.summary()
```

## Step 28: Check target dimensions
This confirms the shape of the one-hot encoded target data.

```python
y.shape
```

## Step 29: Train the model
`fit()` trains the model for 200 epochs using the prepared sequences.

```python
model.fit(x, y, epochs=200)
```

## Step 30: Define test text
This is the starting word provided to the trained model.

```python
test_data="John"
```

## Step 31: Convert test text to numbers
The test word is converted into its tokenizer index.

```python
test_data_index = tokenizer.texts_to_sequences([test_data])
```

## Step 32: Display the test index
This confirms the numeric representation of the test word.

```python
test_data_index
```

## Step 33: Pad the test input
The test sequence is padded to match the input format expected by the model.

```python
padded_test_data = pad_sequences([test_data_index], maxlen = max_len, padding="pre")
```

## Step 34: Generate predictions
The model returns a probability for every word in the vocabulary.

```python
predictions = model.predict(padded_test_data)
```

## Step 35: Find the most likely word
`argmax()` finds the index with the highest prediction probability.

```python
import numpy as np
index = np.argmax(predictions)
index
```

## Step 36: Display prediction confidence
This retrieves the probability assigned to the selected prediction.

```python
value = predictions.flat[index]
value
```

## Step 37: Display all probabilities
This shows the prediction probability for every vocabulary word.

```python
predictions[0]
```

`index_word` converts the predicted numeric index back into readable text.
